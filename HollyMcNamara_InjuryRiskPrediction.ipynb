```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from lifelines import NelsonAalenFitter
```

```python
injury_times = [2018.5, 2022.2, 2023.11]
```

```python
injury_data = pd.DataFrame({
    'duration': injury_times,
    'event': [1] * len(injury_times)
})
```

```python
naf = NelsonAalenFitter()
naf.fit(injury_data['duration'], event_observed = injury_data['event'])
```

```python
plt.figure(figsize = (8, 5))
naf.plot()
plt.title("Nelson-Aalen Cumulative Hazard Estimate of Injury Risk")
plt.xlabel("Year")
plt.ylabel("Cumulative hazard (injury risk)")
plt.grid(True)
plt.show()
```

```python
current_year = 2025.5
predicted_cumulative_hazard = naf.cumulative_hazard_at_times(current_year).values[0]
print(f"Cumulative hazard (injury risk) up to year {current_year}: {predicted_cumulative_hazard:.4f}")
```
